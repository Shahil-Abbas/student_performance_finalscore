Grade Classification Based on Final Score

```python
#1.How can students be categorized into grades based on Final Score?
def assign_grade(score):
    if score >= 85:
        return 'A'
    elif score >= 70:
        return 'B'
    elif score >= 50:
        return 'C'
    else:
        return 'D'

df['Grade'] = df['Final_Score'].apply(assign_grade)
print(df['Grade'].value_counts())
df[['Final_Score','Grade']].head()
```

### 3. Classification Models Applied
- The following models were implemented and evaluated:
  - Logistic Regression

  - Support Vector Machine (SVM)
  - Naive Bayes

  - Random Forest
### 4. Model Evaluation
- Metrics used:
  - Accuracy

  - Balanced and reliable model performance

### 5. Hyperparameter Tuning
